```python
w=[w1,w2,w3]
b=[b1,b2,b3]
hidden_neurons=[]
for w, b in zip(w,b):
    neuron_sum = 0
    for x, weight in zip(i, w):
        neuron_sum += x * weight
    neuron_sum += b
    hidden_neurons.append(neuron_sum)

print("Hidden layer outputs:", hidden_neurons)
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
import numpy as np
```

```python
# Input vector
i = np.array([1, 2, 3, 2.5]).reshape(1, -1)  # shape (1,4)
print("Input shape:", i.shape)
```

```python
# Sequential model
model = Sequential()

# Hidden layer: 3 neurons, no activation (linear), input shape 4
model.add(Dense(3, input_shape=(4,), activation='linear'))

# Output layer: 1 neuron, linear
model.add(Dense(1, activation='linear'))
```

```python
# Hidden layer weights and biases
w_hidden = np.array([
    [0.2, 0.8, -0.5, 1],       # neuron 1 weights
    [0.5, -0.91, 0.26, -0.5],  # neuron 2 weights
    [-0.26, -0.27, 0.17, 0.87] # neuron 3 weights
]).T  # Keras expects shape (input_dim, output_dim)

b_hidden = np.array([2, 3, 0.5])  # biases

# Output layer weights and bias
w_output = np.array([[0.1], [-0.14], [0.5]])  # shape (hidden_neurons, output_neurons)
b_output = np.array([2.5])

# Set weights
model.layers[0].set_weights([w_hidden, b_hidden])
model.layers[1].set_weights([w_output, b_output])
```

```python
# Predict output
hidden_output = model.layers[0](i)  # hidden layer outputs
print("Hidden layer outputs:", hidden_output.numpy())

final_output = model(i)  # full network output
print("Output layer value:", final_output.numpy())
```

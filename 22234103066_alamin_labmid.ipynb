Q1: Write a Python program that can draw a line using both DDA and Bresenham for the
same endpoints (5, 5) → (12, 8).

first cell :DDA part


```python
def dda_line(x1, y1, x2, y2):
    dx = x2 - x1
    dy = y2 - y1
    steps = max(abs(dx), abs(dy))
    x_increment = dx / steps
    y_increment = dy / steps
    x, y = x1, y1
    x_points = []
    y_points = []
    for _ in range(int(steps) + 1):
        x_points.append(round(x))
        y_points.append(round(y))
        x += x_increment
        y += y_increment
    return x_points, y_points

dda_x, dda_y = dda_line(5, 5, 12, 8)
```

second cell Bresenhum

```python
def bresenham_line(x1, y1, x2, y2):
    dx = abs(x2 - x1)
    dy = abs(y2 - y1)
    x_step = 1 if x1 < x2 else -1
    y_step = 1 if y1 < y2 else -1
    x, y = x1, y1
    x_points = []
    y_points = []
    x_points.append(x)
    y_points.append(y)

    if dx > dy:
        p = 2 * dy - dx
        for _ in range(dx):
            x += x_step
            if p < 0:
                p += 2 * dy
            else:
                y += y_step
                p += 2 * dy - 2 * dx
            x_points.append(x)
            y_points.append(y)
    else:
        p = 2 * dx - dy
        for _ in range(dy):
            y += y_step
            if p < 0:
                p += 2 * dx
            else:
                x += x_step
                p += 2 * dx - 2 * dy
            x_points.append(x)
            y_points.append(y)

    return x_points, y_points

bresenham_x, bresenham_y = bresenham_line(5, 5, 12, 8)
```

Q2:Display both lines in the same figure with different colors.

```python
ref_x = [5, 12]
ref_y = [5, 8]

plt.figure(figsize=(8, 8))
plt.plot(dda_x, dda_y, label='DDA', color='yellow', marker='o', linestyle='--')
plt.plot(bresenham_x, bresenham_y, label='Bresenham', color='purple', marker='x', linestyle='--')
plt.plot(ref_x, ref_y, label='Reference Line', color='black', linestyle='-')
plt.legend()
plt.title('DDA vs Bresenham  Drawing')
plt.xlabel('X-coordinate')
plt.ylabel('Y-coordinate')
plt.grid(True)
plt.show()
```

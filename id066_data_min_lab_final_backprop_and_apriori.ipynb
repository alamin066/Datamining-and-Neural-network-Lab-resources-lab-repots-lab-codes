```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def sigmoid(x):
  return 1/(1+np.exp(-x))
```

```python
# Training data
inputs = np.array([[10,20,30],
                   [50,10,20],
                   [60,60,5]])
```

```python
targets = np.array([1, 0, 1])
```

```python
# Hidden neuron 1 weights
w1 = 0.5
w2 = 0.7
w3 = -0.6
# Hidden neuron 2 weights
w4 = 0.9
w5 = 0.4
w6 = 0.8
# Hidden neuron 3 weights
w7 = -1.2
w8 = 0.3
w9 = 1.1
# Output neuron weights
w10 = 0.2
w11 = -0.5
w12 = 0.9
# Hidden layer biases
b1 = 2
b2 = 1.2
b3 = 4.5
 # Output layer bias
b4 = 3.5

learning_rate = 0.001
error_tracking = []
```

```python


for epoch in range(1000):
    total_error = 0
    i = 0

    for x1, x2, x3 in inputs:
        #forward
        h1 = sigmoid(w1*x1 + w2*x2 + w3*x3 + b1)
        h2 = sigmoid(w4*x1 + w5*x2 + w6*x3 + b2)
        h3 = sigmoid(w7*x1 + w8*x2 + w9*x3 + b3)

        o1 = sigmoid(w10*h1 + w11*h2 + w12*h3 + b4)


        target = targets[i]
        total_error += 0.5 * (o1 - target)**2

        #output



        dl_by_dw10 = o1*(1-o1) * h1 * (o1 - target)
        dl_by_dw11 = o1*(1-o1)* h2 * (o1 - target)
        dl_by_dw12 = o1*(1-o1) * h3 * (o1 - target)
        dl_by_db4  = o1*(1-o1)* (o1 - target)

        #Hidden layer gradients


        dl_by_dw1 = h1*(1-h1) * o1*(1-o1) * (o1 - target) * x1
        dl_by_dw2 = h1*(1-h1) * o1*(1-o1) * (o1 - target) * x2
        dl_by_dw3 = h1*(1-h1) * o1*(1-o1) * (o1 - target) * x3
        dl_by_db1 = h1*(1-h1) * o1*(1-o1) * (o1 - target)

        dl_by_dw4 = h2*(1-h2) * o1*(1-o1) * (o1 - target) * x1
        dl_by_dw5 = h2*(1-h2) * o1*(1-o1) * (o1 - target) * x2
        dl_by_dw6 = h2*(1-h2) * o1*(1-o1) * (o1 - target) * x3
        dl_by_db2 = h2*(1-h2) * o1*(1-o1) * (o1 - target)

        dl_by_dw7 = h3*(1-h3) * o1*(1-o1) * (o1 - target) * x1
        dl_by_dw8 = h3*(1-h3)* o1*(1-o1)* (o1 - target)* x2
        dl_by_dw9 = h3*(1-h3) * o1*(1-o1) * (o1 - target) * x3
        dl_by_db3 = h3*(1-h3) * o1*(1-o1) * (o1 - target)

        #  update
        w1 = w1 - learning_rate * dl_by_dw1
        w2 = w2 - learning_rate * dl_by_dw2
        w3 = w3 - learning_rate * dl_by_dw3

        w4 = w4 - learning_rate * dl_by_dw4
        w5 = w5 - learning_rate * dl_by_dw5
        w6 = w6 - learning_rate * dl_by_dw6

        w7 = w7 - learning_rate * dl_by_dw7
        w8 = w8 - learning_rate * dl_by_dw8
        w9 = w9 - learning_rate * dl_by_dw9

        w10 = w10 - learning_rate * dl_by_dw10
        w11 = w11 - learning_rate * dl_by_dw11
        w12 = w12 - learning_rate * dl_by_dw12

        b1 = b1 - learning_rate * dl_by_db1
        b2 = b2 - learning_rate * dl_by_db2
        b3 = b3 - learning_rate * dl_by_db3
        b4 = b4 - learning_rate * dl_by_db4

        target = targets[i]
        total_error += 0.5 * (o1 - target)**2
        i += 1

    error_tracking.append(total_error / 3)
```

```python

plt.plot(range(1000), error_tracking)
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Epoch vs Loss")
plt.show()
```

```python
import numpy as np
import pandas as pd
```

```python
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
data = [['onion', 'beer', 'crisps'],
        ['beer', 'tomato', 'crisps'],
        ['onion', 'crisps'],
        ['beer', 'eggs'],
        ['onion', 'beer'],
        ['onion', 'eggs']]
```

```python
te = TransactionEncoder()
```

```python
te.fit(data)
```

```python
transcation = te.transform(data)
```

```python
df = pd.DataFrame(transcation, columns=te.columns_)
```

```python
df
```

```python
freq_items = apriori(df, min_support=0.4, use_colnames=True)
```

```python
freq_items
```

```python
rules = association_rules(freq_items, metric="confidence", min_threshold=0.6)
```
